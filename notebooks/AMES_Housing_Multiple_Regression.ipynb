# Ames Housing: Multiple Regression

Predict `SalePrice` from `Wood Deck SF` and `Garage Area` using ordinary least squares.

```python
# Generated using Codex CLI
from pathlib import Path
import pandas as pd
import statsmodels.api as sm
```

Load `AmesHousing-1.csv` from the relative `data` folder. The code handles notebooks launched from either the project root or the `notebooks` folder.

```python
csv_candidates = [Path("data/AmesHousing-1.csv"), Path("../data/AmesHousing-1.csv")]
csv_path = next((path for path in csv_candidates if path.exists()), csv_candidates[-1])
if not csv_path.exists():
    raise FileNotFoundError(f"Could not find AmesHousing-1.csv in the relative data directory from {Path.cwd()}")

housing = pd.read_csv(csv_path)
print(f"Loaded {len(housing):,} rows from {csv_path}")
housing.head()
```

## Fit the regression
 
 Rows missing the response or either predictor are excluded. The model includes an intercept. The full regression summary is saved to `results/ames_housing_regression_results.txt`.


```python
target = "SalePrice"
predictors = ["Wood Deck SF", "Garage Area"]
model_data = housing[[target, *predictors]].dropna()

X = sm.add_constant(model_data[predictors])
y = model_data[target]
model = sm.OLS(y, X).fit()

print(f"Rows used: {len(model_data):,} (excluded {len(housing) - len(model_data):,} rows with missing values)")
summary_text = model.summary().as_text()
results_dir = csv_path.parent.parent / "results"
results_dir.mkdir(parents=True, exist_ok=True)
results_path = results_dir / "ames_housing_regression_results.txt"
results_path.write_text(summary_text + "\n", encoding="utf-8")
print(f"Regression results saved to {results_path}")
model.summary()
```

The fitted equation is `SalePrice = intercept + b1 ? Wood Deck SF + b2 ? Garage Area`. The regression summary reports the estimated coefficients and model fit statistics.
