# Ames Housing Descriptive Statistics
 
 This notebook reads `AmesHousing-1.csv` from the relative `data` folder, summarizes its columns and data quality, and writes a human-readable text report to the `results` folder.


```python
# This notebook was produced using Codex CLI.
from pathlib import Path
import pandas as pd

csv_candidates = [Path("data/AmesHousing-1.csv"), Path("../data/AmesHousing-1.csv")]
csv_path = next((path for path in csv_candidates if path.exists()), csv_candidates[-1])
if not csv_path.exists():
    raise FileNotFoundError(f"Could not find AmesHousing-1.csv in the relative data directory from {Path.cwd()}")

df = pd.read_csv(csv_path)
print(f"Loaded {len(df):,} rows and {len(df.columns):,} columns from {csv_path}")
df.head()
```

## Dataset overview

```python
# Data types, non-null counts, and memory usage
df.info()
```

## Numeric descriptive statistics

```python
# Count, mean, standard deviation, quartiles, minimum, and maximum
numeric_summary = df.describe().T
numeric_summary.style.format("{:,.2f}").set_caption("Numeric features")
```

## Categorical descriptive statistics

```python
# Non-null count, distinct values, most frequent value, and its frequency
categorical_summary = df.describe(include=["object"]).T
categorical_summary.style.format({"count": "{:,.0f}", "unique": "{:,.0f}", "freq": "{:,.0f}"}).set_caption("Categorical features")
```

## Missing values

```python
missing = df.isna().sum().sort_values(ascending=False)
missing_summary = missing[missing > 0].to_frame(name="missing_count")
missing_summary["missing_percent"] = (missing_summary["missing_count"] / len(df) * 100).round(2)
missing_summary if not missing_summary.empty else "No missing values found."
```

```python
# Export the summaries as a readable plain-text report
results_dir = csv_path.parent.parent / "results"
results_dir.mkdir(parents=True, exist_ok=True)
report_path = results_dir / "AmesHousing_Descriptive_Statistics.txt"

report_sections = [
    "AMES HOUSING - DESCRIPTIVE STATISTICS",
    "=" * 72,
    f"Dataset: {csv_path.name}",
    f"Rows: {len(df):,}    Columns: {len(df.columns):,}",
    "",
    "NUMERIC FEATURES",
    "-" * 72,
    numeric_summary.to_string(float_format=lambda value: f"{value:,.2f}"),
    "",
    "CATEGORICAL FEATURES",
    "-" * 72,
    categorical_summary.to_string(),
    "",
    "MISSING VALUES",
    "-" * 72,
    missing_summary.to_string() if not missing_summary.empty else "No missing values found.",
]
report_path.write_text("\n".join(report_sections) + "\n", encoding="utf-8")
print(f"Wrote human-readable report to {report_path}")
```
