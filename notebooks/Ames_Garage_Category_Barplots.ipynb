# Ames Housing Garage Category Bar Plots
 
 Create category-count bar plots for `Garage Finish` and `Garage Type`. Each category gets a distinct color, and both PNG files are saved in the shared `results` folder.


```python
# Generated Using Codex CLI
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

csv_candidates = [Path("data/AmesHousing-1.csv"), Path("../data/AmesHousing-1.csv")]
csv_path = next((path for path in csv_candidates if path.exists()), csv_candidates[-1])
if not csv_path.exists():
    raise FileNotFoundError(f"Could not find AmesHousing-1.csv in the relative data directory from {Path.cwd()}")

housing = pd.read_csv(csv_path)
print(f"Loaded {len(housing):,} rows from {csv_path}")
results_dir = csv_path.parent.parent / "results"
results_dir.mkdir(parents=True, exist_ok=True)
```

## Garage Finish

Bars show the number of homes in each category, including missing values if present.

```python
finish_counts = housing["Garage Finish"].fillna("(Missing)").value_counts()
finish_colors = plt.colormaps["tab10"].resampled(len(finish_counts)).colors
fig, ax = plt.subplots(figsize=(8, 5))
ax.bar(finish_counts.index.astype(str), finish_counts.values, color=finish_colors)
ax.set(title="Garage Finish Counts", xlabel="Garage Finish", ylabel="Number of Homes")
ax.tick_params(axis="x", rotation=30)
fig.tight_layout()
finish_plot_path = results_dir / "garage_finish_barplot.png"
fig.savefig(finish_plot_path, dpi=150, bbox_inches="tight")
plt.show()
print(f"Saved {finish_plot_path}")
```

## Garage Type

Bars show the number of homes in each category, including missing values if present.

```python
type_counts = housing["Garage Type"].fillna("(Missing)").value_counts()
type_colors = plt.colormaps["tab10"].resampled(len(type_counts)).colors
fig, ax = plt.subplots(figsize=(8, 5))
ax.bar(type_counts.index.astype(str), type_counts.values, color=type_colors)
ax.set(title="Garage Type Counts", xlabel="Garage Type", ylabel="Number of Homes")
ax.tick_params(axis="x", rotation=30)
fig.tight_layout()
type_plot_path = results_dir / "garage_type_barplot.png"
fig.savefig(type_plot_path, dpi=150, bbox_inches="tight")
plt.show()
print(f"Saved {type_plot_path}")
```
